```python
from run_baselines import experiment_statsmodels, experiment_moirai

result1, _ = experiment_moirai(
                        model_size='small',
                        n_samples=50,
                       output_folder='/raid/yifuc/cache/')
```

```python
result1
```

```python
from collections import defaultdict
from copy import deepcopy
sample_submission = defaultdict(list)
for k, v in result1.items():
    null_list = []
    for score in v:
        score_ = deepcopy(score)
        score_['score'] = 0
        null_list.append(score_)
    sample_submission[k] = null_list
    
```

```python
import pickle 

with open('/raid/yifuc/agent/context-is-key-moirai/solution.pkl', 'wb') as f:
    pickle.dump(result1, f)

with open('/raid/yifuc/agent/context-is-key-moirai/sample_submission.pkl', 'wb') as f:
    pickle.dump(sample_submission, f)
```
